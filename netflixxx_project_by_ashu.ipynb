```python
colms = ['Poster_Url','Overview','Original_Language']
```

```python
df.drop(colms,axis=1,inplace=True,errors='ignore')
```

```python
df.columns
```

```python
df.head()
```

```python
def cat_col(df,col,labels):
    edges =[ df[col].describe()['min'],
         df[col].describe()['25%'],
         df[col].describe()['50%'],
         df[col].describe()['75%'],
         df[col].describe()['max']]
    df[col] = pd.cut(df[col],edges,'Vote_Average',labels=labels,duplicates='drop')
    return df
```

```python
df.columns
```

```python
labels = ['not_popular','below_average','average','popular']
cat_col(df,'Vote_Average',labels)
df['Vote_Average'].unique()
```
